```python
# --- 5. entrenamiento con el dataset si nos dan en el hackaton ---
import os
from ultralytics import YOLO

# 1. Creamos la estructura de carpetas real para tus archivos de entrada
carpetas = [
    '/content/dataset_real/train/images',
    '/content/dataset_real/train/labels',
    '/content/dataset_real/val/images',
    '/content/dataset_real/val/labels'
]

for carpeta in carpetas:
    os.makedirs(carpeta, exist_ok=True)

# 2. Generamos el archivo de configuración YAML apuntando a la ruta real
yolo_real_yaml = """
path: /content/dataset_real
train: train/images
val: val/images
names:
  0: Barco
"""

with open('/content/dataset_real.yaml', 'w') as f:
    f.write(yolo_real_yaml)

print("[OK] Estructura de carpetas creada exitosamente en '/content/dataset_real/'.")
print("[INSTRUCCIÓN] Sube tus imágenes y etiquetas a las carpetas correspondientes antes de ejecutar el entrenamiento de abajo.\n")

def iniciar_entrenamiento_real(epochs=50, batch_size=16, imgsz=640):
    """
    Función lista para entrenar YOLOv8 con tus imágenes reales del Hackatón.
    """
    # Verificación de seguridad para asegurar que el usuario subió imágenes
    imagenes_subidas = os.listdir('/content/dataset_real/train/images')
    if len(imagenes_subidas) == 0:
        print("[AVISO] Aún no has subido imágenes a '/content/dataset_real/train/images'.")
        print("Sube tus archivos para poder iniciar el entrenamiento real. Usando configuración por defecto para demostración.")
        return

    print(f"[INFO] Iniciando Fine-Tuning en GPU T4 con {len(imagenes_subidas)} imágenes reales...")

    # Cargar el modelo base
    model = YOLO('yolov8n.pt')

    # Entrenar modelo con aumentaciones espaciales para satélite
    model.train(
        data='/content/dataset_real.yaml',
        epochs=epochs,
        batch=batch_size,     # Evita CUDA OOM en GPU T4
        imgsz=imgsz,
        device='cuda',        # Forzar GPU
        degrees=180.0,        # Rotación completa 360 grados (satélites)
        flipud=0.5,
        fliplr=0.5,
        mosaic=1.0
    )
    print("[SUCCESS] ¡Entrenamiento completado! Tus mejores pesos están en '/content/runs/detect/train2/weights/best.pt'")

# Descomenta la línea de abajo para iniciar el entrenamiento una vez subas tus imágenes:
# iniciar_entrenamiento_real(epochs=30, batch_size=16, imgsz=640)
```
